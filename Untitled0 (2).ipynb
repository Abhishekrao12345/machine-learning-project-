```python
numeric_feature = data.select_dtypes(include="number")
q1 = numeric_feature.quantile(0.25)
q3 = numeric_feature.quantile(0.75)
qbr = q3 - q1

lower_bound = q1 - 1.5 * qbr
upper_bound = q3 + 1.5 * qbr

outlier = ( numeric_feature < lower_bound) | (numeric_feature > upper_bound)

print(outlier.sum())
```

```python
data = data.drop_duplicates()
```

```python
data['Physical_Activity_Hours'] = data['Physical_Activity_Hours'].clip(lower = 0)
```

```python
data.head()
```

```python
num_col = data.select_dtypes(include="number")
num_col.skew()
```

```python
top_country = data.Country.value_counts().index[:10].tolist()
```

```python
top_country
```

```python
def group_country(country):
  if country in top_country:
    return country
  else :
      return "other_country"
```

```python
data["Grouped_country"] = data['Country'].apply(group_country)
```

```python
data.Grouped_country.value_counts()
```

```python
data.head(1)
```

```python
from sklearn.model_selection import train_test_split

skewd_col = ["Study_Hours"]
oderniary_col = ["Stress_Level"]
numeric_col =["Age", "Avg_Daily_Usage_Hours", "Daily_Unlocks","Physical_Activity_Hours",
                      "Sleep_Hours_Per_Night"]
normal_col = ["Gender", "Academic_Level", "Most_Used_Platform", "Purpose_Of_Use", "Grouped_country"]

feature = skewd_col + oderniary_col + numeric_col + normal_col

x = data[feature]
y = data["Mental_Health_Score"]

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.3, random_state=42)
```

```python
x.head()
```

```python
from sklearn.preprocessing import OrdinalEncoder,OneHotEncoder,StandardScaler,FunctionTransformer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
```

```python
from sklearn.preprocessing import OrdinalEncoder,OneHotEncoder,StandardScaler,FunctionTransformer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline


skew_pipline = Pipeline(steps=[("log_transformer",FunctionTransformer(np.log1p)),
                        ("scale",StandardScaler())
                         ])


numeric_pipline = Pipeline(steps = [("scaler",StandardScaler())
])


ordinary_pipline = Pipeline(steps=[
 ("ordinary",OrdinalEncoder(categories=[['Low', 'Medium', 'High', 'Very High']]))
])


nominal_pipline = Pipeline(steps =[
    ("onehot",OneHotEncoder(handle_unknown="ignore"))
])


preprocess = ColumnTransformer(transformers=[
  ("skew_pipline",skew_pipline,skewd_col),
  ("numeric_pipline",numeric_pipline,numeric_col),
  ("ordinary_pipline",ordinary_pipline,oderniary_col),
  ("nominal_pipline",nominal_pipline,normal_col)
])



```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score , mean_absolute_error , mean_squared_error
```

```python
final_pipline = Pipeline(steps=[
    ("preprocess",preprocess),
    ("model",LinearRegression())
])

final_pipline.fit(x_train,y_train)
predict_test = final_pipline.predict(x_test)
predict_train = final_pipline.predict(x_train)

r2_score_test = r2_score(y_test,predict_test)
r2_score_train = r2_score(y_train,predict_train)
mse_test= mean_squared_error(y_test,predict_test)


print("r2_score_test",r2_score_test)
print("r2_score_trian",r2_score_train)
print("mse",mse_test)
```

```python
from sklearn.ensemble import RandomForestRegressor

final_pipline = Pipeline(steps=[
    ("preprocess",preprocess),
    ("model",RandomForestRegressor())
])

final_pipline.fit(x_train,y_train)
predict_test = final_pipline.predict(x_test)
predict_train = final_pipline.predict(x_train)

r2_score_test = r2_score(y_test,predict_test)
r2_score_train = r2_score(y_train,predict_train)
mse_test = mean_squared_error(y_test,predict_test)

print("r2_score",r2_score_test)
print("r2_score",r2_score_train)
print("mse",mse_test)


```
